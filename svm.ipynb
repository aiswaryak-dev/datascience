```python
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,classification_report
texts=[
    "Red is a bright color",
    "The sky is clear blue",
    "Grass is soft green",
    "The sun shines yellow",
    "Night is dark black",

    "Buses are heavy vechile",
    "Trains ride on an long tracks",
    "Bicycles moves when you pedal",
    "Boat flows on the top of water",
    "Airplanes fly high on the sky"
]
labels=["colors","colors","colors","colors","colors",
       "vehicles","vehicles","vehicles","vehicles","vehicles"]
vectorizer=TfidfVectorizer()
X=vectorizer.fit_transform(texts)
y=labels
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)

model=SVC(kernel="linear")
model.fit(X_train,y_train)
y_pred=model.predict(X_test)
accuracy=accuracy_score(y_test,y_pred)
new_text=["Night is dark black"]
new_text_vector=vectorizer.transform(new_text)
prediction=model.predict(new_text_vector)
print("NewText:",new_text[0])
print("Predicted class:",prediction[0])
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

texts = [
    "Children play in the park",
    "Playing games is fun",
    "The boys play football",
    "The girls play with a ball",
    "Games keep children healthy",
    "Everyone enjoys playing",

    "Music is very enjoyable",
    "Music makes people happy",
    "Children love listening to music",
    "Music can be loud or soft",
    "Songs are a part of music",
    "Music is heard everywhere"
]
labels = [
    "Play", "Play", "Play", "Play", "Play", "Play",
    "Music", "Music", "Music", "Music", "Music", "Music"
]
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(texts)
X_train, X_test, y_train, y_test = train_test_split(
    X, labels, test_size=0.2, random_state=42
)
model = SVC(kernel="linear")
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
new_text = ["Football is a popular game"]
new_text_vector = vectorizer.transform(new_text)
prediction = model.predict(new_text_vector)
print("\nNew Text:", new_text[0])
print("Predicted Class:", prediction[0])
```
